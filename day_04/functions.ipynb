```python
# create a function to add(, sub and yada yada two numbers

def add(a, b):
    return a+b
print(add(3,4))
def sub(a, b):
    return a-b
print(sub(3,4))
# add three numbers
def add(a,b,c):
    return a+b+c
print(add(3,4,5))
```

```python
def add_n(*args):
    print(type(args))
    op=0
    for i in args:
        op+=i
    return op

print(add_n(1,2,3,4))
print(add_n(1,2,3,4,5,6))
```

```python
# global keyword
x=10
def fun(y):
    global x
    x+=y
    return x

print(x)
fun(20)
print(x)
```

```python
#map function

arr=[1,2,3,4]
square=lambda a: a*a

square_array=map(square, arr)
print(square(4))
print(square_array)
print(list(square_array))
```

```python
#filter
#filter even numbers

is_even= lambda x: x%2 == 0
nums = list(range(1,21))
evens = list(filter(is_even, nums))
print(f"evens from 1-20: {evens}")
```

```python
from functools import reduce
#sum of numbers
nums=[1,2,3,4,5]
total = reduce(lambda a, b: a+b, nums)
print(total)
```

```python
# recursion 

def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n*factorial(n-1)
    
factorial(5)
```

```python
#factorial using for loop

def factorial(n):
    op=1
    for i in range(1, n+1):
        op*=i
    return op
print(factorial(5))
```
